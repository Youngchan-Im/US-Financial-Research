```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from dotenv import load_dotenv

print("pandas_version:", pd.__version__)
print("numpy_version:", np.__version__)
print("matplotlib_ready:", True)
print("dotenv_ready:", callable(load_dotenv))
```

```python
company_candidates = pd.DataFrame(
    {
       "company_name": [
           "Apple Inc.",
           "Microsoft Corporation",
       ],
       "ticker": ["AAPL", "MSFT"],
       "research_reason": [
           "제품과 서비스가 함께 있는 기업 구조를 살펴보기 좋다",
           "소프트웨어와 클라우드 중심 사업 구조를 살펴보기 쉽다",
       ],
       "planned_data_checks": [3, 3]
       
    }
)

candidate_count = len(company_candidates)
print("candidate_count:", candidate_count)
company_candidates
```

```python
ax = company_candidates.plot(
    kind="bar",
    x="ticker",
    y="planned_data_checks",
    legend=False,
)
ax.set_title("Planned data cekcs by ticker")
ax.set_xlabel("ticker")
ax.set_ylabel("planned_data_ckecks")
plt.show()
```
