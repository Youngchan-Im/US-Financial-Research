```python
review_plan = []

for company in company_candidates:
    for metric_name in metric_names:
        review_plan.append(
            {
                "ticker": company["ticker"],
                "company_name": company["company_name"],
                "metric_name": metric_name,
            }
        )

review_task_count = len(review_plan)
print("review_task_count:", review_task_count)
review_plan[:5]
```

```python
def build_company_summary(company):
    ticker = company["ticker"]
    company_name = company["company_name"]
    planned_count = company["planned_data_checks"]
    return f"{ticker} - {company_name} | planned_data_checks={planned_count}"

for company in company_candidates:
    print(build_company_summary(company))
```

```python
def count_review_tasks(companies, metrics):
    return len(companies) * len(metrics)

expected_review_taks_count = count_review_tasks(company_candidates, metric_names)

print("expected_review_taks_count:", expected_review_taks_count)
print("review_task_count:", review_task_count)
print("count_match:", expected_review_taks_count == review_task_count)
```
