```python
project_name = "U.S. Financial Research Project"
session_number = 3
notebook_name = "01_python_basics.ipynb"

print("project_name:", project_name)
print("session_number:", session_number)
print("notebook_name:", notebook_name)
```

```python
ticker_list = ["AAPL", "MSFT", "NVDA"]
metric_names = ["revenue", "operating_income", "net_income"]
planned_data_checks = len(metric_names)

print("ticker_list:", ticker_list)
print("metric_names:", metric_names)
print("planned_data_checks:", planned_data_checks)
```

```python
apple_record = {
    "company_name": "Apple Inc.",
    "ticker": "AAPL",
    "research_reason": "제품과 서비스가 함께 있는 기업 구조를 살펴보기 좋다.",
    "planned_data_checks": planned_data_checks,
}

print("company_name:", apple_record["company_name"])
print("ticker:", apple_record["ticker"])
print("planned_data_checks", apple_record["planned_data_checks"])
```

```python
company_candidates = [
    {
        "company_name": "Apple Inc.",
        "ticker": "APPL",
        "research_reason": "제품과 서비스가 함께 있는 기업 구조를 살펴보기 좋다.",
        "planned_data_checks": planned_data_checks,
    },
    {
        "company_name": "Microsoft Corporation",
        "ticker": "MSFT",
        "research_reason": "소프트웨어와 클라우드 중심 사업 구조를 살펴보기 좋다.",
        "planned_data_checks": planned_data_checks,
    },
    {
        "company_name": "NVIDIA Corporation",
        "ticker": "NVDA",
        "research_reason": "반도체와 데이터센터 수요 관련 자료를 살펴보기 쉽다.",
        "planned_data_checks": planned_data_checks,
    },
]

candidate_count = len(company_candidates)
print("candidate_count:", candidate_count)
company_candidates
```

```python
for company in company_candidates:
    ticker = company["ticker"]
    company_name = company["company_name"]
    planned_count = company["planned_data_checks"]
    print(f"{ticker}: {company_name} | planned_data_checks={planned_count}")
```

```python
review_plan = []

for company in company_candidates:
    for metric_name in metric_names:
        review_plan.append(
            {
                "ticker": company["ticker"],
                "company_name": company["company_name"],
                "metric_name": metric_name,
            }
        )

review_task_count = len(review_plan)
print("review_task_count:", review_task_count)
review_plan[:5]
```

```python
def build_company_summary(company):
    ticker = company["ticker"]
    company_name = company["company_name"]
    planned_count = company["planned_data_checks"]
    return f"{ticker} - {company_name} | planned_data_checks={planned_count}"

for company in company_candidates:
    print(build_company_summary(company))
```

```python
def count_review_tasks(companies, metrics):
    return len(companies) * len(metrics)

expected_review_taks_count = count_review_tasks(company_candidates, metric_names)

print("expected_review_taks_count:", expected_review_taks_count)
print("review_task_count:", review_task_count)
print("count_match:", expected_review_taks_count == review_task_count)
```
